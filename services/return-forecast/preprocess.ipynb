```python
import pandas as pd

file_path_2021 = '/content/2021 Data.xls'

# Load all sheets from the 2021 Data.xls file
# sheet_name=None returns a dictionary of DataFrames, one for each sheet
all_sheets_2021 = pd.read_excel(file_path_2021, sheet_name=None, header=3)

processed_dfs = []

for sheet_name, df_sheet in all_sheets_2021.items():
    # Ensure the sheet is not empty and contains enough rows for header processing
    if not df_sheet.empty and len(df_sheet) > 0:
        # Set the column names using the first row of the DataFrame (which contains the actual headers)
        # Make a copy to avoid SettingWithCopyWarning later
        df_sheet_copy = df_sheet.copy()
        df_sheet_copy.columns = df_sheet_copy.iloc[0]

        # Drop the first row as it has now become the header
        df_sheet_copy = df_sheet_copy[1:].reset_index(drop=True)

        # Clean column names by stripping whitespace
        df_sheet_copy.columns = df_sheet_copy.columns.str.strip()

        # Convert the 'TRADING DATE' column to datetime objects, coercing errors
        # Only attempt conversion if 'TRADING DATE' column exists
        if 'TRADING DATE' in df_sheet_copy.columns:
            df_sheet_copy['TRADING DATE'] = pd.to_datetime(df_sheet_copy['TRADING DATE'], errors='coerce')

            # Drop rows where 'TRADING DATE' is NaT (if any)
            df_sheet_copy.dropna(subset=['TRADING DATE'], inplace=True)

            # Convert the 'TRADING DATE' column to ISO 8601 (YYYY-MM-DD) string format
            df_sheet_copy['TRADING DATE'] = df_sheet_copy['TRADING DATE'].dt.strftime('%Y-%m-%d')
            processed_dfs.append(df_sheet_copy)

# Concatenate all processed DataFrames into a single df_2021
df_2021 = pd.concat(processed_dfs, ignore_index=True)

print(f"Processed file: {file_path_2021} from all sheets")
print("DataFrame after date conversion and concatenation:")
display(df_2021.head())
df_2021.info()
```

```python
import numpy as np

# --- Adjust for known corporate actions (stock splits / share consolidations) ---
# These are NOT data errors — they're real, documented corporate actions that change
# the share basis without changing shareholder value. Left unadjusted, they show up
# as fake, extreme "returns" that will corrupt Daily_Return, Log_Return, MACD, RSI,
# ROC, and the 5-day-ahead target right around the event date.
#
# price_multiplier = factor to multiply PRE-event prices by, to bring them onto the
# POST-event share basis (i.e. adjust history to match today's share count).
#   > 1  -> a consolidation happened (fewer, higher-priced shares)
#   < 1  -> a split happened (more, lower-priced shares)
#
# Add more (company, effective_date, price_multiplier) rows here if you confirm
# additional corporate actions for the other banks later.
corporate_actions = [
    ('ABL',  '2024-07-15', 10.0),   # Amana Bank: 10 old shares consolidated into 1 new share
    ('SAMP', '2021-03-23', 1/3),    # Sampath Bank: 1 old share split into 3 new shares
]

price_cols  = ['OPEN PRICE (Rs.)', 'PRICE HIGH (Rs.)', 'PRICE LOW (Rs.)', 'CLOSE PRICE (Rs.)']
volume_cols = ['TRADE VOLUME (No.)', 'SHARE VOLUME (No.)']
# TURNOVER (Rs.) is deliberately left untouched — it's the actual Rs. value traded
# that day, which a split/consolidation does not change.

for company, effective_date, price_multiplier in corporate_actions:
    effective_date = pd.to_datetime(effective_date)
    mask = (combined_annual_df['COMPANY ID'] == company) & (combined_annual_df['Date'] < effective_date)

    combined_annual_df.loc[mask, price_cols]  = combined_annual_df.loc[mask, price_cols] * price_multiplier
    combined_annual_df.loc[mask, volume_cols] = combined_annual_df.loc[mask, volume_cols] / price_multiplier

    print(f"Adjusted {mask.sum()} pre-{effective_date.date()} rows for {company} "
          f"(prices x{price_multiplier:.4f}, volumes x{1/price_multiplier:.4f})")
```

```python
check = combined_annual_df.sort_values(['COMPANY ID','Date']).copy()
check['ret_check'] = check.groupby('COMPANY ID')['CLOSE PRICE (Rs.)'].pct_change()
remaining = check[check['ret_check'].abs() > 0.4][['COMPANY ID','Date','CLOSE PRICE (Rs.)','ret_check']]
print("Remaining extreme single-day moves (>40%) after adjustment:")
display(remaining)
```

```python
import numpy as np

# Ensure 'Date' is datetime type for proper sorting and shifting
# (It should already be from the previous step, but good to ensure)
combined_annual_df['Date'] = pd.to_datetime(combined_annual_df['Date'])

# Sort by company and then by date to ensure correct 'previous day' logic
combined_annual_df.sort_values(by=['COMPANY ID', 'Date'], inplace=True)

# --- Handle missing 'OPEN PRICE (Rs.)' ---
initial_missing_open = combined_annual_df['OPEN PRICE (Rs.)'].isnull().sum()

# Calculate previous day's close price within each company group
combined_annual_df['PREV_CLOSE'] = combined_annual_df.groupby('COMPANY ID')['CLOSE PRICE (Rs.)'].shift(1)

# Fill missing 'OPEN PRICE (Rs.)' with 'PREV_CLOSE'
combined_annual_df['OPEN PRICE (Rs.)'] = combined_annual_df['OPEN PRICE (Rs.)'].fillna(combined_annual_df['PREV_CLOSE'])

# Drop the temporary 'PREV_CLOSE' column
combined_annual_df.drop(columns=['PREV_CLOSE'], inplace=True)

filled_open_count = initial_missing_open - combined_annual_df['OPEN PRICE (Rs.)'].isnull().sum()
print(f"Filled {filled_open_count} missing 'OPEN PRICE (Rs.)' values with previous day's close.")

# --- Handle missing 'PRICE LOW (Rs.)' ---
initial_missing_low = combined_annual_df['PRICE LOW (Rs.)'].isnull().sum()

# Fill missing 'PRICE LOW (Rs.)' with the minimum of 'OPEN PRICE (Rs.)' and 'CLOSE PRICE (Rs.)'
# It's important that 'OPEN PRICE (Rs.)' has already been partially imputed here.
# Use np.fmin for element-wise minimum, handling potential NaN values gracefully
combined_annual_df['PRICE LOW (Rs.)'] = combined_annual_df.apply(
    lambda row: np.fmin(row['OPEN PRICE (Rs.)'], row['CLOSE PRICE (Rs.)'])
                if pd.isnull(row['PRICE LOW (Rs.)'])
                else row['PRICE LOW (Rs.)'], axis=1
)

filled_low_count = initial_missing_low - combined_annual_df['PRICE LOW (Rs.)'].isnull().sum()
print(f"Filled {filled_low_count} missing 'PRICE LOW (Rs.)' values with min(Open, Close).")

# Display summary of changes and info
print("\nDataFrame after imputing missing Open and Low prices:")
display(combined_annual_df.head())
display(combined_annual_df.tail())
combined_annual_df.info()
```

### Merging All Cleaned Data into a Single Panel

Now that all individual datasets (`combined_annual_df`, `filtered_bank_data`, `filtered_exchange_rate_data`, and `filtered_ccpi_df`) are cleaned and preprocessed (including the one-month forward shift for inflation data to avoid lookahead bias), we will merge them into a single panel. To correctly handle the differing publication schedules of these financial indicators, we will use a nearest-prior-value join. This means for each stock's trading date, we will fetch the most recent available value from the market index, FX rate, and inflation datasets *prior to or on* that trading date.

```python
# 1. Rename 'TRADING DATE' in combined_annual_df to 'Date' for consistency
combined_annual_df.rename(columns={'TRADING DATE': 'Date'}, inplace=True)

# Ensure all 'Date' columns are datetime type, which is a prerequisite for merge_asof
combined_annual_df['Date'] = pd.to_datetime(combined_annual_df['Date'])
filtered_bank_data['Date'] = pd.to_datetime(filtered_bank_data['Date'])
filtered_exchange_rate_data['Date'] = pd.to_datetime(filtered_exchange_rate_data['Date'])
# filtered_ccpi_df['Date'] is already datetime and shifted from previous step

print("Renamed 'TRADING DATE' to 'Date' in combined_annual_df and ensured all 'Date' columns are datetime type.")

# 2. Sort all DataFrames by their 'Date' column
# This is crucial for pd.merge_asof with direction='backward'
combined_annual_df.sort_values(by='Date', inplace=True)
filtered_bank_data.sort_values(by='Date', inplace=True)
filtered_exchange_rate_data.sort_values(by='Date', inplace=True)
filtered_ccpi_df.sort_values(by='Date', inplace=True)

print("All DataFrames sorted by 'Date'.")

# 3. Perform nearest-prior-value merges using pd.merge_asof

# Start with combined_annual_df as the left DataFrame
merged_df = combined_annual_df.copy()

# Merge with filtered_bank_data
merged_df = pd.merge_asof(merged_df,
                     filtered_bank_data,
                     on='Date',
                     direction='backward',
                     suffixes=('', '_Bank')) # Use suffixes to avoid column name clashes
print("Merged with filtered_bank_data using nearest-prior-value join.")

# Merge with filtered_exchange_rate_data
merged_df = pd.merge_asof(merged_df,
                     filtered_exchange_rate_data,
                     on='Date',
                     direction='backward',
                     suffixes=('', '_FX')) # Use suffixes
print("Merged with filtered_exchange_rate_data using nearest-prior-value join.")

# Merge with filtered_ccpi_df
merged_df = pd.merge_asof(merged_df,
                     filtered_ccpi_df,
                     on='Date',
                     direction='backward',
                     suffixes=('', '_Inflation')) # Use suffixes
print("Merged with filtered_ccpi_df using nearest-prior-value join.")

print("\nFinal merged DataFrame head (showing first 10 rows for better context):")
display(merged_df.head(10))

print("\nFinal merged DataFrame tail (showing last 10 rows):")
display(merged_df.tail(10))

print("\nFinal merged DataFrame info:")
merged_df.info()

print("\nChecking for any remaining NaNs in the newly merged columns:")
print(merged_df[['Sector_Index', 'Exchange Rate', 'YoY_Inflation_%']].isnull().sum())
```

### Feature Engineering: Technical Indicators (Per Stock)

To prepare the dataset for predictive modeling, we will create several technical indicators. It's critical that these calculations are performed *separately for each company* to prevent data leakage and ensure that the indicators reflect the specific stock's performance.

We will compute:

1.  **Returns:** Daily Return and Log Return.
2.  **Price Structure:** High-Low Range relative to Close Price, and Open-Close Change.
3.  **Trend Indicators:** 5-day, 10-day, and 20-day Simple Moving Averages (SMA) and Exponential Moving Averages (EMA) of the Close Price. Additionally, a ratio of Close Price to its 20-day EMA will be calculated.

```python
import numpy as np

# Ensure the DataFrame is sorted by COMPANY ID and Date for correct rolling/shifting operations
merged_df.sort_values(by=['COMPANY ID', 'Date'], inplace=True)

# Group by 'COMPANY ID' to perform calculations independently for each stock
merged_df_grouped = merged_df.groupby('COMPANY ID')

# 1. Returns
# Daily Return: (Close Price_t - Close Price_{t-1}) / Close Price_{t-1}
merged_df['Daily_Return'] = merged_df_grouped['CLOSE PRICE (Rs.)'].pct_change()

# Log Return: log(Close Price_t / Close Price_{t-1})
merged_df['Log_Return'] = np.log(merged_df_grouped['CLOSE PRICE (Rs.)'].ffill() / merged_df_grouped['CLOSE PRICE (Rs.)'].ffill().shift(1))

# 2. Price Structure
# High-Low Range relative to Close
merged_df['High_Low_Range'] = (merged_df['PRICE HIGH (Rs.)'] - merged_df['PRICE LOW (Rs.)']) / merged_df['CLOSE PRICE (Rs.)']

# Open-Close Change
merged_df['Open_Close_Change'] = (merged_df['CLOSE PRICE (Rs.)'] - merged_df['OPEN PRICE (Rs.)']) / merged_df['OPEN PRICE (Rs.)']

# 3. Trend Indicators
# Define periods for moving averages
ma_periods = [5, 10, 20]

for period in ma_periods:
    # Simple Moving Average (SMA)
    merged_df[f'SMA_{period}'] = merged_df_grouped['CLOSE PRICE (Rs.)'].transform(lambda x: x.rolling(window=period).mean())

    # Exponential Moving Average (EMA)
    # Adjust=False is often preferred for consistency with older EMA definitions and stability
    merged_df[f'EMA_{period}'] = merged_df_grouped['CLOSE PRICE (Rs.)'].transform(lambda x: x.ewm(span=period, adjust=False).mean())

# Price-vs-20-day-average ratio (using EMA_20 as it's more responsive)
merged_df['Price_vs_EMA20_Ratio'] = merged_df['CLOSE PRICE (Rs.)'] / merged_df['EMA_20']

print("Feature engineering complete. Displaying head and info of the DataFrame with new features:")
display(merged_df.head(15))
merged_df.info()

print("\nChecking for NaNs in newly created feature columns:")
new_features = ['Daily_Return', 'Log_Return', 'High_Low_Range', 'Open_Close_Change'] + \
               [f'SMA_{p}' for p in ma_periods] + [f'EMA_{p}' for p in ma_periods] + \
               ['Price_vs_EMA20_Ratio']
print(merged_df[new_features].isnull().sum())
```

### Feature Engineering: Momentum and Oscillator Indicators

```python
def calculate_rsi(series, window=14):
    delta = series.diff()
    gain = delta.where(delta > 0, 0)
    loss = -delta.where(delta < 0, 0)

    avg_gain = gain.ewm(com=window - 1, adjust=False).mean()
    avg_loss = loss.ewm(com=window - 1, adjust=False).mean()

    rs = avg_gain / avg_loss
    rsi = 100 - (100 / (1 + rs))
    return rsi

merged_df['RSI_14'] = merged_df_grouped['CLOSE PRICE (Rs.)'].transform(calculate_rsi)

print("RSI_14 calculated.")
# Display first few rows with new feature and check for NaNs
display(merged_df[['COMPANY ID', 'Date', 'CLOSE PRICE (Rs.)', 'RSI_14']].head(20))
print(merged_df[['RSI_14']].isnull().sum())
```

```python
macd_fast_period = 12
macd_slow_period = 26
macd_signal_period = 9

# Calculate EMA fast for each group using transform
ema_fast_grouped = merged_df_grouped['CLOSE PRICE (Rs.)'].transform(lambda x: x.ewm(span=macd_fast_period, adjust=False).mean())

# Calculate EMA slow for each group using transform
ema_slow_grouped = merged_df_grouped['CLOSE PRICE (Rs.)'].transform(lambda x: x.ewm(span=macd_slow_period, adjust=False).mean())

# Calculate MACD line
merged_df['MACD'] = ema_fast_grouped - ema_slow_grouped

# Calculate MACD Signal line based on the newly created MACD line, grouped again
# Re-grouping by 'COMPANY ID' for the signal line ensures that the EMA for the signal line is calculated correctly per company.
merged_df['MACD_Signal'] = merged_df.groupby('COMPANY ID')['MACD'].transform(lambda x: x.ewm(span=macd_signal_period, adjust=False).mean())

print("MACD and MACD_Signal calculated.")
# Display first few rows with new features and check for NaNs
display(merged_df[['COMPANY ID', 'Date', 'CLOSE PRICE (Rs.)', 'MACD', 'MACD_Signal']].head(30))
print(merged_df[['MACD', 'MACD_Signal']].isnull().sum())
```

```python
bollinger_window = 20
bollinger_num_std = 2

# Calculate Middle Band (20-day SMA)
merged_df['BB_Middle'] = merged_df_grouped['CLOSE PRICE (Rs.)'].transform(lambda x: x.rolling(window=bollinger_window).mean())

# Calculate 20-day rolling standard deviation
merged_df['BB_StdDev'] = merged_df_grouped['CLOSE PRICE (Rs.)'].transform(lambda x: x.rolling(window=bollinger_window).std())

# Calculate Upper and Lower Bands
merged_df['BB_Upper'] = merged_df['BB_Middle'] + (merged_df['BB_StdDev'] * bollinger_num_std)
merged_df['BB_Lower'] = merged_df['BB_Middle'] - (merged_df['BB_StdDev'] * bollinger_num_std)

# Calculate %B
merged_df['BB_Percent_B'] = (merged_df['CLOSE PRICE (Rs.)'] - merged_df['BB_Lower']) / (merged_df['BB_Upper'] - merged_df['BB_Lower'])

print("Bollinger Bands (Middle, Upper, Lower, %B) calculated.")
# Display first few rows with new features and check for NaNs
display(merged_df[['COMPANY ID', 'Date', 'CLOSE PRICE (Rs.)', 'BB_Middle', 'BB_Upper', 'BB_Lower', 'BB_Percent_B']].head(25))
print(merged_df[['BB_Middle', 'BB_Upper', 'BB_Lower', 'BB_Percent_B']].isnull().sum())
```

```python
roc_period = 10
merged_df['ROC_10Day'] = merged_df_grouped['CLOSE PRICE (Rs.)'].transform(lambda x: ((x / x.shift(roc_period)) - 1) * 100)

print("10-day Rate of Change calculated.")
# Display first few rows with new feature and check for NaNs
display(merged_df[['COMPANY ID', 'Date', 'CLOSE PRICE (Rs.)', 'ROC_10Day']].head(15))
print(merged_df[['ROC_10Day']].isnull().sum())
```

```python
import numpy as np # Ensure numpy is imported for np.where

stochastic_k_window = 14
stochastic_d_window = 3

def calculate_stochastic(df_group, k_window, d_window):
    lowest_low = df_group['PRICE LOW (Rs.)'].rolling(window=k_window).min()
    highest_high = df_group['PRICE HIGH (Rs.)'].rolling(window=k_window).max()

    # Avoid division by zero and handle potential inf/nan values carefully
    range_hl = highest_high - lowest_low

    # Calculate %K using np.where to handle cases where range_hl is zero or NaN
    # Ensure the result is a Series with the original index
    k_percent = pd.Series(np.where(
        (range_hl == 0) | range_hl.isna(),
        0.0, # Assign 0.0 if range is 0 or NaN, or choose np.nan if that's preferred for division by zero
        ((df_group['CLOSE PRICE (Rs.)'] - lowest_low) / range_hl) * 100
    ), index=df_group.index)

    # Calculate %D based on %K
    d_percent = k_percent.rolling(window=d_window).mean()

    # Return a DataFrame with both K and D, maintaining the original group index
    return pd.DataFrame({'Stochastic_K': k_percent, 'Stochastic_D': d_percent}, index=df_group.index)

# Apply the function to each group. The .apply() method on GroupBy will correctly handle
# a function that returns a DataFrame, concatenating the results.
stochastic_results_df = merged_df_grouped.apply(lambda x: calculate_stochastic(x, stochastic_k_window, stochastic_d_window), include_groups=False)

# Assign the results back to the merged_df, explicitly aligning indices by dropping the 'COMPANY ID' level
merged_df['Stochastic_K'] = stochastic_results_df['Stochastic_K'].droplevel(level=0)
merged_df['Stochastic_D'] = stochastic_results_df['Stochastic_D'].droplevel(level=0)

print("Stochastic %K and %D calculated.")
# Display first few rows with new features and check for NaNs
display(merged_df[['COMPANY ID', 'Date', 'CLOSE PRICE (Rs.)', 'PRICE HIGH (Rs.)', 'PRICE LOW (Rs.)', 'Stochastic_K', 'Stochastic_D']].head(20))
print(merged_df[['Stochastic_K', 'Stochastic_D']].isnull().sum())
```

### Feature Engineering: Market & Macro Context

```python
import numpy as np

# Re-group the DataFrame by 'COMPANY ID' to ensure calculations are company-specific
merged_df_grouped = merged_df.groupby('COMPANY ID')

# 1. Sector Index Return
# Calculate day-over-day change in the banking index, grouped by company ID
# Since Sector_Index is constant for a given date across companies, this correctly calculates daily market return.
merged_df['Sector_Index_Return'] = merged_df_grouped['Sector_Index'].pct_change()

# 2. USD/LKR Return
# Calculate day-over-day change in the exchange rate, grouped by company ID
# Similar to Sector_Index_Return, Exchange Rate is constant for a given date across companies.
merged_df['USD_LKR_Return'] = merged_df_grouped['Exchange Rate'].pct_change()

# 3. Relative Strength
# How much a stock's return beats or lags the sector that day
merged_df['Relative_Strength'] = merged_df['Daily_Return'] - merged_df['Sector_Index_Return']

# 4. Rolling 20-day Beta
beta_window = 20

# Calculate rolling covariance of Daily_Return with Sector_Index_Return for each company
# Using .apply(lambda x: ...) to ensure the result aligns with the original DataFrame's index correctly
merged_df['Rolling_Covariance'] = merged_df_grouped.apply(
    lambda x: x['Daily_Return'].rolling(window=beta_window).cov(x['Sector_Index_Return']),
    include_groups=False # Added to resolve DeprecationWarning
).droplevel(0)

# Calculate rolling variance of Sector_Index_Return for each company
merged_df['Rolling_Variance_Sector'] = merged_df_grouped['Sector_Index_Return'].transform(
    lambda x: x.rolling(window=beta_window).var()
)

# Calculate Beta, handling potential division by zero or zero variance
# Replace inf/-inf values that can result from division by zero/very small numbers with NaN
merged_df['Rolling_20Day_Beta'] = (
    merged_df['Rolling_Covariance'] / merged_df['Rolling_Variance_Sector']
).replace([np.inf, -np.inf], np.nan)

# Drop temporary covariance and variance columns
merged_df.drop(columns=['Rolling_Covariance', 'Rolling_Variance_Sector'], inplace=True)

print("Market & Macro context features calculated.")
# Display first few rows with new features and check for NaNs
display(merged_df[[
    'COMPANY ID', 'Date', 'Daily_Return', 'Sector_Index', 'Sector_Index_Return',
    'Exchange Rate', 'USD_LKR_Return', 'Relative_Strength', 'Rolling_20Day_Beta'
]].head(25))

print("\nChecking for NaNs in newly created Market & Macro context feature columns:")
market_macro_features = ['Sector_Index_Return', 'USD_LKR_Return', 'Relative_Strength', 'Rolling_20Day_Beta']
print(merged_df[market_macro_features].isnull().sum().sort_values(ascending=False))
```

### Feature Engineering: Calendar Features

```python
import numpy as np

# Ensure 'Date' column is datetime type
merged_df['Date'] = pd.to_datetime(merged_df['Date'])

# 1. Cyclical encoding of Day of Week
# Day of week: Monday=0, Sunday=6
merged_df['Day_of_Week'] = merged_df['Date'].dt.dayofweek
merged_df['Day_of_Week_Sin'] = np.sin(2 * np.pi * merged_df['Day_of_Week'] / 7)
merged_df['Day_of_Week_Cos'] = np.cos(2 * np.pi * merged_df['Day_of_Week'] / 7)

# 2. Cyclical encoding of Month
# Month: January=1, December=12
merged_df['Month'] = merged_df['Date'].dt.month
merged_df['Month_Sin'] = np.sin(2 * np.pi * merged_df['Month'] / 12)
merged_df['Month_Cos'] = np.cos(2 * np.pi * merged_df['Month'] / 12)

# 3. Quarter-end-month flag
# Quarters typically end in March (3), June (6), September (9), December (12)
merged_df['Is_Quarter_End_Month'] = merged_df['Month'].isin([3, 6, 9, 12]).astype(int)

print("Calendar features calculated.")
# Display first few rows with new features and check for NaNs
display(merged_df[[
    'COMPANY ID', 'Date', 'Day_of_Week', 'Day_of_Week_Sin', 'Day_of_Week_Cos',
    'Month', 'Month_Sin', 'Month_Cos', 'Is_Quarter_End_Month'
]].head(25))

print("\nChecking for NaNs in newly created Calendar features columns:")
calendar_features = ['Day_of_Week', 'Day_of_Week_Sin', 'Day_of_Week_Cos', 'Month', 'Month_Sin', 'Month_Cos', 'Is_Quarter_End_Month']
print(merged_df[calendar_features].isnull().sum().sort_values(ascending=False))
```

### Target Variable Creation: 5-Day-Ahead Price and Return

Now, we will create the target variables required for predictive modeling. We need to define:

1.  **5-Day-Ahead Close Price (`Target_Close_5Day_Ahead`):** This is the closing price of a stock 5 trading days into the future. This will serve as a direct regression target.
2.  **5-Day-Ahead Return (`Target_Return_5Day_Ahead`):** This represents the percentage change in the closing price over the next 5 trading days. This is often a more robust target for forecasting as it normalizes price movements.

These calculations will be performed per company to maintain the integrity of the time series and prevent data leakage.

```python
merged_df_grouped = merged_df.groupby('COMPANY ID')

# Calculate Close price and Return targets for every horizon from t+1 to t+5
for h in range(1, 6):
    merged_df[f'Target_Close_{h}Day_Ahead'] = merged_df_grouped['CLOSE PRICE (Rs.)'].shift(-h)
    merged_df[f'Target_Return_{h}Day_Ahead'] = (merged_df[f'Target_Close_{h}Day_Ahead'] / merged_df['CLOSE PRICE (Rs.)']) - 1

print("Target variables created for horizons t+1 through t+5.")

target_close_cols = [f'Target_Close_{h}Day_Ahead' for h in range(1, 6)]
target_return_cols = [f'Target_Return_{h}Day_Ahead' for h in range(1, 6)]

display(merged_df[['COMPANY ID', 'Date', 'CLOSE PRICE (Rs.)'] + target_return_cols].tail(10))

print("\nChecking for NaNs in newly created target columns:")
print(merged_df[target_close_cols + target_return_cols].isnull().sum().sort_values(ascending=False))
```

### Step 11: Check for Redundant Features (High Correlation)

To further streamline our dataset and mitigate multicollinearity, we will examine the correlations between all engineered features. Features that are highly correlated (e.g., absolute correlation coefficient > 0.95) often provide redundant information, and one can be removed without significant loss of predictive power. This step helps in model interpretability and can prevent issues in some machine learning algorithms.

```python
import numpy as np

# Exclude non-feature columns (identifiers, date) and target variables
features_to_check = [col for col in merged_df.columns if col not in [
    'COMPANY ID', 'Date',
    'Target_Close_5Day_Ahead', 'Target_Return_5Day_Ahead'
]]

# Create a DataFrame with only the features
features_df = merged_df[features_to_check].copy()

# Calculate the correlation matrix
correlation_matrix = features_df.corr().abs()

# Select upper triangle of correlation matrix to avoid duplicates and self-correlation
upper_tri = correlation_matrix.where(np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool))

# Find features with correlation greater than 0.95
highly_correlated_pairs = []
for i in range(len(upper_tri.columns)):
    for j in range(i):
        if upper_tri.iloc[i, j] > 0.95:
            highly_correlated_pairs.append((upper_tri.columns[i], upper_tri.columns[j], upper_tri.iloc[i, j]))

if not highly_correlated_pairs:
    print("No highly correlated feature pairs (above 0.95) found.")
else:
    print("Highly correlated feature pairs (absolute correlation > 0.95):")
    for f1, f2, corr_val in highly_correlated_pairs:
        print(f"- {f1} and {f2}: {corr_val:.4f}")

    # Based on domain knowledge and common practices, select features to drop
    # Prioritize keeping more direct or widely used indicators
    features_to_drop = []

    # Example decisions based on typical financial feature engineering:
    # Price components: Keep CLOSE PRICE, drop others if highly correlated
    if ('PRICE HIGH (Rs.)', 'CLOSE PRICE (Rs.)', correlation_matrix.loc['PRICE HIGH (Rs.)', 'CLOSE PRICE (Rs.)']) in highly_correlated_pairs:
        if 'PRICE HIGH (Rs.)' not in features_to_drop: features_to_drop.append('PRICE HIGH (Rs.)')
    if ('PRICE LOW (Rs.)', 'CLOSE PRICE (Rs.)', correlation_matrix.loc['PRICE LOW (Rs.)', 'CLOSE PRICE (Rs.)']) in highly_correlated_pairs:
        if 'PRICE LOW (Rs.)' not in features_to_drop: features_to_drop.append('PRICE LOW (Rs.)')
    if ('OPEN PRICE (Rs.)', 'CLOSE PRICE (Rs.)', correlation_matrix.loc['OPEN PRICE (Rs.)', 'CLOSE PRICE (Rs.)']) in highly_correlated_pairs:
        if 'OPEN PRICE (Rs.)' not in features_to_drop: features_to_drop.append('OPEN PRICE (Rs.)')

    # Moving Averages: EMA is often preferred for responsiveness; BB_Middle is an SMA
    if 'SMA_20' in features_to_check and 'EMA_20' in features_to_check and correlation_matrix.loc['SMA_20', 'EMA_20'] > 0.95:
        if 'SMA_20' not in features_to_drop: features_to_drop.append('SMA_20')
    if 'SMA_10' in features_to_check and 'EMA_10' in features_to_check and correlation_matrix.loc['SMA_10', 'EMA_10'] > 0.95:
        if 'SMA_10' not in features_to_drop: features_to_drop.append('SMA_10')
    if 'SMA_5' in features_to_check and 'EMA_5' in features_to_check and correlation_matrix.loc['SMA_5', 'EMA_5'] > 0.95:
        if 'SMA_5' not in features_to_drop: features_to_drop.append('SMA_5')

    # Bollinger Bands: BB_Middle is effectively an SMA; BB_Upper/Lower are derivatives.
    # BB_Percent_B is a normalized oscillator and often more useful.
    if 'BB_Middle' in features_to_check and 'EMA_20' in features_to_check and correlation_matrix.loc['BB_Middle', 'EMA_20'] > 0.95:
        if 'BB_Middle' not in features_to_drop: features_to_drop.append('BB_Middle')
    if 'BB_Upper' in features_to_check and 'CLOSE PRICE (Rs.)' in features_to_check and correlation_matrix.loc['BB_Upper', 'CLOSE PRICE (Rs.)'] > 0.95:
        if 'BB_Upper' not in features_to_drop: features_to_drop.append('BB_Upper')
    if 'BB_Lower' in features_to_check and 'CLOSE PRICE (Rs.)' in features_to_check and correlation_matrix.loc['BB_Lower', 'CLOSE PRICE (Rs.)'] > 0.95:
        if 'BB_Lower' not in features_to_drop: features_to_drop.append('BB_Lower')
    if 'BB_StdDev' in features_to_check and 'Vol_20Day' in features_to_check and correlation_matrix.loc['BB_StdDev', 'Vol_20Day'] > 0.95:
        # Both measure 20-day volatility. Keep Vol_20Day as it's a more direct measure of return std dev.
        if 'BB_StdDev' not in features_to_drop: features_to_drop.append('BB_StdDev')

    # Volume/Turnover: Turnover captures both price and volume.
    if 'SHARE VOLUME (No.)' in features_to_check and 'TURNOVER (Rs.)' in features_to_check and correlation_matrix.loc['SHARE VOLUME (No.)', 'TURNOVER (Rs.)'] > 0.95:
        if 'SHARE VOLUME (No.)' not in features_to_drop: features_to_drop.append('SHARE VOLUME (No.)')

    # Stochastic: D is a smoothed K
    if 'Stochastic_K' in features_to_check and 'Stochastic_D' in features_to_check and correlation_matrix.loc['Stochastic_K', 'Stochastic_D'] > 0.95:
        # Keep D as it's typically used as the signal line and is smoother
        if 'Stochastic_K' not in features_to_drop: features_to_drop.append('Stochastic_K')

    # Calendar features: Keep sin/cos, drop raw numeric
    if 'Day_of_Week' in features_to_check and ('Day_of_Week_Sin' in features_to_check or 'Day_of_Week_Cos' in features_to_check):
        if 'Day_of_Week' not in features_to_drop: features_to_drop.append('Day_of_Week')
    if 'Month' in features_to_check and ('Month_Sin' in features_to_check or 'Month_Cos' in features_to_check):
        if 'Month' not in features_to_drop: features_to_drop.append('Month')

    # MACD related: MACD_Signal is derived from MACD
    if 'MACD_Signal' in features_to_check and 'MACD' in features_to_check and correlation_matrix.loc['MACD_Signal', 'MACD'] > 0.95:
        # Keeping MACD and MACD_Signal is standard for MACD crossover strategies, but if one must be dropped for high corr, keep MACD.
        # However, for MACD, typically both are kept for crossover signals. Let's not drop one of them right now unless the correlation is extremely high and causes issues.
        pass # Decide to keep both for now, unless specific instruction to drop one.

    # Remove duplicates from the list of features to drop
    features_to_drop = list(set(features_to_drop))

    if features_to_drop:
        print(f"\nRecommended features to drop due to high correlation: {features_to_drop}")
        merged_df.drop(columns=features_to_drop, inplace=True)
        print(f"Dropped {len(features_to_drop)} features.")
    else:
        print("\nNo features recommended for dropping based on current criteria or manual selection.")

print("\nShape of DataFrame after correlation-based feature reduction:", merged_df.shape)
print("\nFinal DataFrame head after feature reduction:")
display(merged_df.head())
print("\nFinal DataFrame info after feature reduction:")
merged_df.info()
```

### Step 12: Split and Scale Data


```python
from sklearn.preprocessing import StandardScaler

train_end_date = pd.to_datetime('2023-12-31')
val_end_date = pd.to_datetime('2024-12-31')

train_df = merged_df[merged_df['Date'] <= train_end_date]
val_df = merged_df[(merged_df['Date'] > train_end_date) & (merged_df['Date'] <= val_end_date)]
test_df = merged_df[merged_df['Date'] > val_end_date]

print(f"Training set shape: {train_df.shape}")
print(f"Validation set shape: {val_df.shape}")
print(f"Test set shape: {test_df.shape}")

# Exclude identifiers AND every target column (any horizon), not just the original two
features_cols = [col for col in merged_df.columns
                  if col not in ['COMPANY ID', 'Date'] and not col.startswith('Target_')]
target_cols = [f'Target_Return_{h}Day_Ahead' for h in range(1, 6)]  # t+1 ... t+5, in order

X_train = train_df[features_cols]
y_train = train_df[target_cols]
X_val = val_df[features_cols]
y_val = val_df[target_cols]
X_test = test_df[features_cols]
y_test = test_df[target_cols]

company_id_train = train_df['COMPANY ID']
company_id_val = val_df['COMPANY ID']
company_id_test = test_df['COMPANY ID']

scalers = {}
scaled_X_train_list, scaled_X_val_list, scaled_X_test_list = [], [], []

for company_id in merged_df['COMPANY ID'].unique():
    X_train_company = X_train[company_id_train == company_id]
    X_val_company = X_val[company_id_val == company_id]
    X_test_company = X_test[company_id_test == company_id]

    if not X_train_company.empty:
        scaler = StandardScaler()
        scaler.fit(X_train_company)
        scalers[company_id] = scaler

        def attach_ids(scaled_array, source_df, index):
            out = pd.DataFrame(scaled_array, columns=features_cols, index=index)
            out.insert(0, 'Date', source_df.loc[index, 'Date'].values)
            out.insert(0, 'COMPANY ID', company_id)
            return out

        scaled_X_train_list.append(attach_ids(scaler.transform(X_train_company), train_df, X_train_company.index))
        if not X_val_company.empty:
            scaled_X_val_list.append(attach_ids(scaler.transform(X_val_company), val_df, X_val_company.index))
        if not X_test_company.empty:
            scaled_X_test_list.append(attach_ids(scaler.transform(X_test_company), test_df, X_test_company.index))
    else:
        print(f"Warning: Company {company_id} has no training data. Skipping.")

X_train_scaled = pd.concat(scaled_X_train_list).sort_index()
X_val_scaled = pd.concat(scaled_X_val_list).sort_index()
X_test_scaled = pd.concat(scaled_X_test_list).sort_index()

y_train = y_train.loc[X_train_scaled.index].copy()
y_val = y_val.loc[X_val_scaled.index].copy()
y_test = y_test.loc[X_test_scaled.index].copy()

print(f"X_train_scaled: {X_train_scaled.shape}, y_train: {y_train.shape}")
print(f"X_val_scaled: {X_val_scaled.shape}, y_val: {y_val.shape}")
print(f"X_test_scaled: {X_test_scaled.shape}, y_test: {y_test.shape}")
X_train_scaled.head()
```
